## Vector Stores (aka. Vector Databases)

* Store embeddings in a very fast searchable database.

```python
from langchain_community.document_loaders import TextLoader
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_text_splitters import CharacterTextSplitter
from langchain_chroma import Chroma


# 1. Load document
loaded_document = TextLoader(
    "data/state of the union.txt",
    encoding="utf-8"
).load()


# 2. Split document
text_splitter = CharacterTextSplitter(chunk_size=1000,chunk_overlap=0)

chunks_of_text = text_splitter.split_documents(loaded_document)


# 3. Create embedding model
embeddings_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")


```

```python
len(chunks_of_text)
```

```python
# 4. Create Chroma vector database
vector_db = Chroma.from_documents(
    documents=chunks_of_text,
    embedding=embeddings_model,
    collection_name="state_of_union",
    persist_directory="./chroma_db"
)
```

```python
vector_db
```

```python
question = "What did the president say about the John Lewis Rights Act?"

response = vector_db.similarity_search(question)

print(response[0].page_content)
```

## Vector Store as Retriever
* Find the embedding that best answers your question

```python
from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import FAISS
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_text_splitters import CharacterTextSplitter
from langchain_chroma import Chroma

loaded_document = TextLoader("data/state of the union.txt", encoding="utf-8").load()

text_splitter = CharacterTextSplitter(chunk_size=1000,chunk_overlap=0)

chunks_of_text = text_splitter.split_documents(loaded_document)

embeddings_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

vector_db = FAISS.from_documents(chunks_of_text, embeddings_model)
```

```python
vector_db
```

```python
retriever = vector_db.as_retriever(search_kwargs={"k":2})
```

```python
response = retriever.invoke("What did he say about ketanji brown jackson?")
response
```
