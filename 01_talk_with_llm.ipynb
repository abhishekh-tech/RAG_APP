```python
import os 
from dotenv import load_dotenv,find_dotenv
_= load_dotenv(find_dotenv())
groq_api_key = os.environ["GROQ_API_KEY"]
```

```python
# Completion model

from langchain_groq import ChatGroq

llmModel = ChatGroq(
    model="qwen/qwen3.8-27b",
    temperature=0
)
```

```python
response = llmModel.invoke("Tell me one fun fact about the Kennedy family.")
```

```python
print(response.content)
```

```python
for chunk in llmModel.stream(
    "Tell me one fun fact about Narendra Modi. i need detailed response"
):
    print(chunk.content,end="",flush=True)
```

```python
from langchain_core.messages import(
    SystemMessage,
    HumanMessage,
    AIMessage
)

# Different types of messages
messages = [
    SystemMessage(
        content="You are a helpful AI teacher. Explain concepts simply."
    ),

    HumanMessage(
        content="What is a neural network?"
    ),

    AIMessage(
        content="A neural network is a machine learning model inspired by the human brain."
    ),

    HumanMessage(
        content="What is a neuron in a neural network?"
    )
]
```

```python
response = creativellmModel.invoke(messages)
print(response.content)
```

```python
response = llmModel.invoke(messages)
print(response.content)
```
